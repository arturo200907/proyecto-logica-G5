```python
g=Grafos()
```

```python
def filtro(dict_tseitin, regla): 
    h = inorder_to_tree(regla) 
    letras = h.letras()
    for letter in dict_tseitin.copy(): 
        if letter not in letras: 
            del dict_tseitin[letter] 

    return dict_tseitin 
```

## Regla 1

```python
%%time 
#REGLA 1
A1 = g.reglas[0]
ts1 = tseitin(A1)   
r1 = walkSAT(ts1) 
I1 = filtro(r1[1], A1) 
```
